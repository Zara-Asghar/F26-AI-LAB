```python

import re

texts = {
    "Punctuation": "Hello, how are you? I am fine!",
    "Contraction": "I can't attend today's class.",
    "Email": "Contact me at student@gmail.com",
    "Identifier": "My course code is CS4944",
    "Empty String": "",
    "Urdu Text": "آپ کیسے ہیں؟ میں ٹھیک ہوں۔"
}

for name, text in texts.items():
    whitespace_tokens = text.split()
    regex_tokens = re.findall(r"\b\w+\b", text, flags=re.UNICODE)

    print("\nScenario:", name)
    print("Original Text:", text)
    print("Whitespace Tokens:", whitespace_tokens)
    print("Whitespace Count:", len(whitespace_tokens))
    print("Regex Tokens:", regex_tokens)
    print("Regex Count:", len(regex_tokens))
```
