```python

from difflib import get_close_matches

faqs = {
    "What is tokenization?": {
        "answer": "Tokenization means breaking text into smaller parts called tokens.",
        "source": "Lab 6 Manual - Tokenization"
    },
    "What is chunking?": {
        "answer": "Chunking means dividing a large document into smaller parts while keeping useful information together.",
        "source": "Lab 6 Manual - Text Chunking"
    }
}

print("Welcome to the FAQ Chatbot!")
print("Ask a question or type 'exit' to quit.")

while True:
    user_question = input("\nEnter your question: ").strip()

    if user_question.lower() == "exit":
        print("Goodbye!")
        break

    if user_question == "":
        print("Please enter a question.")
        continue

    matched = False

    for question, details in faqs.items():
        if user_question.lower() == question.lower():
            print("\nAnswer:", details["answer"])
            print("Source:", details["source"])
            matched = True
            break

    if not matched:
        suggestions = get_close_matches(
            user_question, faqs.keys(), n=2, cutoff=0.3
        )

        if suggestions:
            print("\nDid you mean:")
            for i, question in enumerate(suggestions, 1):
                print(i, question)

            choice = input("Enter the question number to confirm, or 0 to cancel: ")

            if choice.isdigit() and 1 <= int(choice) <= len(suggestions):
                selected = suggestions[int(choice) - 1]
                print("\nAnswer:", faqs[selected]["answer"])
                print("Source:", faqs[selected]["source"])
            else:
                print("Please contact the help desk for further assistance.")
        else:
            print("No matching question found. Please contact the help desk.")
```
