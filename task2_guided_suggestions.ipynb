```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

# Documents
documents = [
    "Students can request an extension if they are sick. They must provide a medical certificate as evidence.",
    "Customers should report damaged parcels within 48 hours and provide photos of the damaged item.",
    "When an API token expires, the client should use a refresh token to get a new access token."
]

# Chunk IDs
chunk_ids = ["DOC1_CH1", "DOC2_CH1", "DOC3_CH1"]

# Queries
queries = [
    "How can a student get extra time after illness?",
    "What should I do if my parcel arrives damaged?",
    "How can I get a new API token?"
]

# TF-IDF setup
vectorizer = TfidfVectorizer()
tfidf_data = vectorizer.fit_transform(documents)

# Embedding setup
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
embeddings = model.encode(documents)

# Search using both methods
for query in queries:

    print("\nQuery:", query)

    # TF-IDF
    query_data = vectorizer.transform([query])
    scores = cosine_similarity(query_data, tfidf_data)[0]

    print("\nTF-IDF Results:")
    for i in scores.argsort()[::-1][:3]:
        print(chunk_ids[i], documents[i], "Score:", round(scores[i], 3))

    # Embeddings
    query_embedding = model.encode([query])
    scores = cosine_similarity(query_embedding, embeddings)[0]

    print("\nEmbedding Results:")
    for i in scores.argsort()[::-1][:3]:
        print(chunk_ids[i], documents[i], "Score:", round(scores[i], 3))
```
